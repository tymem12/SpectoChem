```python
from torch.utils.data import DataLoader
from gjepa.datasets.cv_vis.tmqmg_star import TMQMGStarDataset
```

```python
dataset = TMQMGStarDataset(
    target_cols=['homo_lumo_gap_gasphase']  # predict HOMO-LUMO gap
)

print("X:", dataset.X.shape, "y:", dataset.y.shape)

loader = DataLoader(dataset, batch_size=64, shuffle=True)

# Iterate through one batch
for X_batch, y_batch in loader:
    print(X_batch.shape, y_batch.shape)
    break
```

```python
dataset = TMQMGStarDataset(
    target_cols=['homo_lumo_gap_gasphase'],  # predict HOMO-LUMO gap
    atom_filter=3 # 3rd block
)

print("X:", dataset.X.shape, "y:", dataset.y.shape)

loader = DataLoader(dataset, batch_size=64, shuffle=True)

# Iterate through one batch
for X_batch, y_batch in loader:
    print(X_batch.shape, y_batch.shape)
    break
```
